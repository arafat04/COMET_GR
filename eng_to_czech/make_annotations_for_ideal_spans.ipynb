Les bons scores d’accord inter-annotateurs sont présentés et analysés plus en détail.
```python
import random

def create_shifted_annotations(spans, offsets):
    shifted_annotations = []
    
    for i, sentence_spans in enumerate(spans):
        sent_offsets = offsets[i]
        new_sentence_annotations = []
        
        for span_dict in sentence_spans:
            orig_start = span_dict['startnew']
            orig_end = span_dict['endnew']
            
            start_token_idx = None
            end_token_idx = None
            
            for idx, (s, e) in enumerate(sent_offsets):
                if s <= orig_start < e:
                    start_token_idx = idx
                if s < orig_end <= e:
                    end_token_idx = idx
            
            if start_token_idx is None or end_token_idx is None:
                start_token_idx = 0
                end_token_idx = len(sent_offsets) - 1
            
            annotations_for_this_span = []
            for _ in range(2):
                which_end = random.choice(['start', 'end'])
                direction = random.choice(['L', 'R'])
                step = random.choice([1, 2, 3, 4])
                
                new_start_idx = start_token_idx
                new_end_idx = end_token_idx
                
                if which_end == 'start':
                    if direction == 'L':
                        new_start_idx = max(0, start_token_idx - step)
                    else:
                        new_start_idx = min(len(sent_offsets) - 1, start_token_idx + step)
                    if new_start_idx > end_token_idx:
                        new_start_idx = end_token_idx
                else:
                    if direction == 'L':
                        new_end_idx = max(0, end_token_idx - step)
                    else:
                        new_end_idx = min(len(sent_offsets) - 1, end_token_idx + step)
                    if new_end_idx < start_token_idx:
                        new_end_idx = start_token_idx
                
                new_start_char = sent_offsets[new_start_idx][0]
                new_end_char = sent_offsets[new_end_idx][1]
                
                if new_start_char < orig_end and orig_start < new_end_char:
                    if which_end == 'start':
                        if direction == 'L':
                            alt_start_idx = min(len(sent_offsets) - 1, start_token_idx + step)
                            if alt_start_idx > end_token_idx:
                                alt_start_idx = end_token_idx
                        else:
                            alt_start_idx = max(0, start_token_idx - step)
                        alt_start_char = sent_offsets[alt_start_idx][0]
                        alt_end_char = sent_offsets[end_token_idx][1]
                        if alt_start_char < orig_end and orig_start < alt_end_char:
                            final_start_char = sent_offsets[start_token_idx][0]
                            final_end_char = sent_offsets[end_token_idx][1]
                        else:
                            final_start_char = alt_start_char
                            final_end_char = alt_end_char
                    else:
                        if direction == 'L':
                            alt_end_idx = min(len(sent_offsets) - 1, end_token_idx + step)
                            if alt_end_idx < start_token_idx:
                                alt_end_idx = start_token_idx
                        else:
                            alt_end_idx = max(0, end_token_idx - step)
                        alt_start_char = sent_offsets[start_token_idx][0]
                        alt_end_char = sent_offsets[alt_end_idx][1]
                        if alt_start_char < orig_end and orig_start < alt_end_char:
                            final_start_char = sent_offsets[start_token_idx][0]
                            final_end_char = sent_offsets[end_token_idx][1]
                        else:
                            final_start_char = alt_start_char
                            final_end_char = alt_end_char
                else:
                    final_start_char = new_start_char
                    final_end_char = new_end_char
                
                annotations_for_this_span.append({
                    'startnew': final_start_char,
                    'endnew': final_end_char
                })
            
            new_sentence_annotations.append(annotations_for_this_span)
        
        shifted_annotations.append(new_sentence_annotations)
    
    return shifted_annotations

# Example usage:

shifted_annots = create_shifted_annotations(filtered_spans_min_4_char, updated_word_offsets_935)
```
