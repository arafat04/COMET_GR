**Get sentence length for each abstract**

```python
def sentence_length_abstract(translations):
    sentences_length_of_abstract = []
    for idx, value in translations.items():
        sentences_length_of_abstract.append(len(value))
    return sentences_length_of_abstract
```

```python
def get_sentences_len(first_4_text_list,df_en_cz, source_or_translation):
    import ast
    sentences_length_of_all_abstract = []
    
    if source_or_translation == "source":
        column_name = 'source'
        which_len = "document_length_source"
    else:
        column_name = 'translation'
        which_len = "document_length_translation"
    print(column_name)
    print(which_len)
    for abstract_id in first_4_text_list:
        row = filtered_df_first_4[filtered_df_first_4['translation_id'] == abstract_id]
        indices = ast.literal_eval(row["sentence_indices"].iloc[0])
        first_index = indices[0]
        last_index = indices[-1]
        print(abstract_id, first_index, last_index)

        sentences_length_of_each_abstract = sentence_length_abstract(df_en_cz.iloc[first_index:last_index+1][f"{column_name}"])
        sentences_length_of_all_abstract.append({
            "translation_id":abstract_id,
            "translation_starting_index": first_index,
            "translation_ending_index": last_index,
            f"{which_len}":sentences_length_of_each_abstract
        })
    return sentences_length_of_all_abstract
```

### get the relative position P of the span in the text (the start index of the span divided by the total text length)

```python
def get_relative_postion_P_of_spans(spans, abstract_length):
    relative_positions_P = []
    for index, span in enumerate(spans):
        start_offset = span[0]
        relative_position = start_offset / abstract_length
        relative_positions_P.append(round(relative_position,3))
    return relative_positions_P
```

### now make the annotation table using 4 formulas

```python
def create_annotation_rank_4(spans, L, P, translation_id):
    # severity is doubled
    severity_label_values = {
        'minor': .24,
        'major': .48,
        'critical': .96,
    }
    annotation_rank_4 = []
    for index, span in enumerate(spans):
        severity = span[3]
        #print("severity: ", severity)
        severity_value = severity_label_values[severity]
        #print(severity_value)
        #print("severity value doubled: ", 2 * severity_value)
        confidence = span[4]
        #print("confidence: ", confidence)
        relative_length = L[index]
        #print("relative_length: ", relative_length)
        relative_position = P[index]
        #print("relative_position: ", relative_position)

        rank = confidence + 2 * severity_value + relative_length + relative_position
        #print("rank: ", rank)
        annotation_rank_4.append(rank)
    #print("annotation_rank: ", annotation_rank_4)
    #print("annotation_rank sorted: ", sorted(annotation_rank_4, reverse=True))
    # Zip ranks with tuples
    combined = list(zip(annotation_rank_4, spans))

    # Sort by rank in descending order
    sorted_combined = sorted(combined, key=lambda x: x[0], reverse=True)

    # Unpack back into ranks and tuples
    sorted_annotation_rank_4, sorted_spans_4 = map(list, zip(*sorted_combined))

    print("Sorted Ranks:", sorted_annotation_rank_4)
    print("Sorted Tuples:", sorted_spans_4)
    return sorted_spans_4
```

```python
def populate_annotations(annotations):
    import json
    import mysql.connector
    from mysql.connector import Error

    # Database configuration
    DB_CONFIG = {
        'host': 'localhost',
        'user': 'root',
        'password': '',
        'database': 'matos',
        'port': 3306
    }

    # JSON file path

    def populate_annotations():
        try:
            # Load JSON data
        
            
            # Connect to database
            connection = mysql.connector.connect(**DB_CONFIG)
            cursor = connection.cursor()
            
            # SQL insert query for Annotation table
            insert_query = """
            INSERT INTO Annotation (
                annotation_rank, 
                translation_id, 
                annotator_id, 
                start_span, 
                end_span, 
                severity_level, 
                confidence
            ) VALUES (%s, %s, %s, %s, %s, %s, %s)
            """
            
            # Prepare data for insertion
            records = []
            for ann in annotations:
                records.append((
                    ann['annotation_rank'],
                    ann['translation_id'],
                    ann['annotator_id'],
                    ann['start_span'],
                    ann['end_span'],
                    ann['severity_level'],
                    ann['confidence']
                ))
            
            # Execute batch insert
            cursor.executemany(insert_query, records)
            connection.commit()
            print(f"Successfully inserted {len(records)} annotations")
            
        except Error as e:
            print(f"Database error: {e}")
            if 'connection' in locals() and connection.is_connected():
                connection.rollback()
        except Exception as e:
            print(f"General error: {e}")
        finally:
            if 'connection' in locals() and connection.is_connected():
                cursor.close()
                connection.close()

    if __name__ == "__main__":
        populate_annotations()
```
